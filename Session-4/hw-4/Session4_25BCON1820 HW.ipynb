## Program 1: Even-Odd Number — C to Python Translation

```python

num = int(input("Enter a number: "))

if num % 2 == 0:
    print(num, "is even")
else:
    print(num, "is odd")
```

Program 2: Sum of Natural Numbers


```python

n = int(input("Enter the value of n: "))
total = 0

for i in range(1, n + 1):
    total += i

print("Sum of natural numbers =", total)
```

Program 3: Prime Number


```python

n = int(input("Enter a number: "))

if n < 2:
    print(n, "is not a prime number")
else:
    is_prime = True

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        print(n, "is a prime number")
    else:
        print(n, "is not a prime number")
```

**Output Verification:** For input 7, the program checks whether any number from 2 up to the square root of 7 divides it exactly. Since no such divisor exists, 7 is identified as a prime number.

## Mismatch Analysis and Fixes

The Python programs were tested using sample inputs. The outputs shown above are the expected outputs for those inputs. They must be compared with the actual outputs of the corresponding C programs.

1. **Even-Odd:** The number is checked using the modulo operator (%). The condition works similarly in C and Python.
2. **Sum of Natural Numbers:** Python's `range(1, n + 1)` includes numbers from 1 to n because the ending value of `range()` is excluded.
3. **Prime Number:** The program checks possible divisors and stops when a divisor is found. The variable `is_prime` stores the result of the check.

## C vs Python: Cheat Sheet

| No. | Feature | C | Python |
|---:|---|---|---|
| 1 | Variable declaration | Data type is declared | Type is inferred at runtime |
| 2 | Input | `scanf()` | `input()` |
| 3 | Output | `printf()` | `print()` |
| 4 | Statement ending | Usually uses semicolons | Semicolons are generally unnecessary |
| 5 | Code blocks | Uses braces `{}` | Uses indentation |
| 6 | Loop syntax | `for` syntax includes initialization, condition and update | `for` can iterate over `range()` |
| 7 | Conditions | Uses `if` and `else` with parentheses commonly | Uses `if` and `else` followed by a colon |
| 8 | Data grouping | Can use `struct` | Can use dictionaries or classes |
